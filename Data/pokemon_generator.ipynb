```python
import pandas as pd

df = pd.read_csv("pokemon_all_gens.csv")
df["type2"] = df["type2"].fillna("None")


def choose_canonical_form(group):
    """
    Select the best 'default' row for one pokedex_id.
    """
    preferred_suffixes = [
        "",
        "-normal",
        "-ordinary",
        "-standard",
        "-altered",
        "-incarnate",
        "-land",
        "-overcast",
        "-plant",
        "-west",
        "-male",
        "-aria",
        "-red-striped",
        "-ice-face",
        "-disguised",
        "-average",
        "-one-segment",
        "-curly",
        "-zero",
        "-a"
    ]
    species = str(group["species_name"].iloc[0])
    exact = group[group["name"] == species]
    if not exact.empty:
        return exact.iloc[0]
    for suffix in preferred_suffixes[1:]:
        target = f"{species}{suffix}"
        match = group[group["name"] == target]
        if not match.empty:
            return match.iloc[0]
    unwanted_keywords = [
        "mega", "gmax", "totem", "origin", "therian", "sky", "zen", "resolute", "pirouette", "school",
        "blade", "busted", "amped", "low-key", "antique", "phony", "ruby-cream", "matcha-cream", "mint-cream",
        "lemon-cream", "salted-cream", "ruby-swirl", "caramel-swirl", "rainbow-swirl", "hangry", "rapid-strike",
        "single-strike", "hero", "family-of-four", "family-of-three", "three-segment", "roaming", "bloodmoon"
    ]
    cleaned = group.copy()
    cleaned["bad_form"] = cleaned["name"].apply(
        lambda x: any(word in str(x) for word in unwanted_keywords)
    )
    safer = cleaned[cleaned["name"] == False]
    if not safer.empty:
        return safer.iloc[0]

canonical_df = (
    df.groupby("pokedex_id", group_keys=False)
    .apply(choose_canonical_form)
    .reset_index()
)

print(canonical_df.shape[0])
print(canonical_df["pokedex_id"].nunique())
```
